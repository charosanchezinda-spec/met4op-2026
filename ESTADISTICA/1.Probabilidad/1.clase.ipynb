### Vínculo entre frecuencia relativa y probabilidad

$p_n$ (o $\hat{p}_n$) representa la proporción de resultados $= 1$ a medida que se tira el dado ($eje\ x$)

- A medida que se incrementa la $n$, la proporción de $1$ sobre el total converge a $p = \frac{1}{6}$, que es la probabilidad de obtener un $1$
- Esta convergencia se fundamenta matemáticamente por la ***Ley de los Grandes Números (LGN)***


```python
import numpy as np

# Ley de los Grandes Números: la frecuencia relativa de sacar un "1" en un dado
# converge a la probabilidad teórica 1/6 ≈ 0.1667 a medida que crecen los tiros.
rng = np.random.default_rng(88)   # semilla fija -> resultado reproducible

for n in [1,2,3,4,5,6, 10, 100, 1_000, 10_000, 100_000]:
    tiros = rng.integers(1, 7, size=n)          # dados entre 1 y 6
    frec_relativa = np.mean(tiros == 1)
    print(f"n = {n:>7,}  ->  frecuencia relativa de '1' = {frec_relativa:.4f}")

print(f"\nProbabilidad teórica = 1/6 = {1/6:.4f}")
```

### Definiciones de Probabilidad

- **Experimento aleatorio**:
 proceso con resultado incierto. Puede repetirse en iguales condiciones para la obtención de "datos".

- **Resultados**: Realizaciones del experimento aleatorio. No conocemos con seguridad el resultado, pero sí podemos conocer la **probabilidad de ocurrencia** de cada resultado.
- **Espacio muestral:** conjunto de *todos los posibles resultados* de un experimento aleatorio.
- **Evento:** algún subconjunto del espacio muestral.

### Experimento aleatorio vs Experimento social

- **Experimento aleatorio:** tirar un dado, una moneda, sacar una carta de un mazo. *Todos estos experimentos pueden repetirse bajo las mismas condiciones*. Si bien conocemos cuales son sus posibles resultados, no es posible decir con exactitud cuál es el resultado al ejecutarlo.

- **Experimento social:** Evaluar la probabilidad de ocurrencia de cierto evento social es **independiente** del contexto y de lo social. Por esa razón, ***no es*** un experimento que se pueda repetir bajo las mismas condiciones. Existe una *Probabilidad Condicional*, ejemplo: ¿Qué probabilidad existe de que se produzca una huelga laboral en Argentina?

----------------------------------------------------

- **Frecuencia relativa:** se basa en observar *qué tan seguido* ocurre un evento en la práctica, a través de las repeticiones del experimento.

Para la huelga, podemos considerar los datos históricos:

Supongamos que en los últimos 120 meses hubo 12 huelgas.

La frecuencia relativa sería: 12/120=0,11 → “probabilidad histórica” de que haya huelga en un mes cualquiera = 10%.

Limitación: Esto asume que las condiciones son comparables, pero sabemos que factores económicos y sociales varían mes a mes, por lo que la probabilidad real es condicional a esas circunstancias.

#### 3. Diferencia entre experimento aleatorio y experimento social

#### Experimento aleatorio (sentido estadístico)

Es un procedimiento **controlado y reproducible** en el que el azar determina el resultado. La aleatoriedad es **formal**: proviene de un mecanismo
de sorteo o selección probabilística definido por el investigador.

| Característica | Descripción |
|:--|:--|
| Control | El investigador controla el procedimiento |
| Reproducibilidad | Se puede repetir en condiciones idénticas |
| Fuente del azar | Mecanismo de selección (sorteo, muestra aleatoria) |
| Espacio muestral | Conocido y fijo de antemano |
| Objetivo | Describir o estimar parámetros poblacionales |

**Ejemplos:**
- Sortear 1.200 ciudadanos del padrón para una encuesta
- Asignar aleatoriamente municipios a un programa
  de presupuesto participativo para evaluarlo
- Seleccionar al azar legisladores para integrar comisiones

#### Experimento social (sentido politológico)

Es una situación real del mundo político en la que **intervienen decisiones humanas, contextos históricos e instituciones**. La aleatoriedad no es formal: el resultado depende de la interacción de múltiples factores, muchos de ellos no controlables.

| Característica | Descripción |
|:--|:--|
| Control | Muy limitado o nulo |
| Reproducibilidad | Casi imposible: el contexto cambia |
| Fuente del azar | Incertidumbre política, decisiones humanas |
| Espacio muestral | Difícil de delimitar con exactitud |
| Objetivo | Explicar causas, predecir comportamientos |

**Ejemplos:**
- Una elección presidencial
- La decisión de un partido de formar una coalición
- El resultado de una votación legislativa reñida
- El impacto de una crisis económica en el apoyo al gobierno

### Teoría de conjuntos

#### 1. Conjunto

Un conjunto es una colección bien definida de elementos. 

**Ejemplos:**

$$A = \{\text{UCR}, \text{PJ}, \text{PRO}, \text{FIT}, \text{MID}\}$$
$$B = \{\text{Buenos Aires}, \text{Córdoba}, \text{Santa Fe}\}$$
$$C = \{\text{Diputados que votaron a favor de la reforma}\}$$

#### 2. Subconjunto (⊆)

A es subconjunto de B si **todo elemento de A también pertenece a B**.

$$A \subseteq B \iff \forall x \in A \Rightarrow x \in B$$

**Ejemplo:**
Sea el universo U = todos los partidos con representación en el Congreso Nacional.

$$U = \{\text{UCR}, \text{PJ}, \text{PRO}, \text{FIT}, \text{MID}, \text{Libres del Sur}\}$$
$$A = \{\text{UCR}, \text{PJ}, \text{PRO}\} \quad \text{(partidos con más de 20 bancas)}$$

Como UCR, PJ y PRO pertenecen a U:

$$A \subseteq U$$

#### 3. Unión (A ∪ B)

<figure style="text-align:center; margin:1em 0;">
  <img src="media/union.png"  width="520">
  <figcaption><em>Union</em></figcaption>
</figure>

La unión contiene **todos los elementos que pertenecen a A, a B, o a ambos**.

$$A \cup B = \{x \mid x \in A \text{ o } x \in B\}$$

**Ejemplo:**
$$A = \{\text{legisladores que votaron la reforma previsional}\}$$
$$B = \{\text{legisladores que votaron la reforma electoral}\}$$

$$A \cup B = \{\text{legisladores que votaron al menos una de las dos reformas}\}$$

Si A tiene 180 legisladores y B tiene 140,
y 90 votaron ambas:

$$|A \cup B| = 180 + 140 - 90 = 230$$

#### 4. Intersección (A ∩ B)

<figure style="text-align:center; margin:1em 0;">
  <img src="media/interseccion.png"  width="520">
  <figcaption><em>Union</em></figcaption>
</figure>

La intersección contiene **solo los elementos que
pertenecen simultáneamente a A y a B**.

$$A \cap B = \{x \mid x \in A \text{ y } x \in B\}$$

**Ejemplo:**
$$A = \{\text{provincias con más del 40\% de pobreza}\}$$
$$B = \{\text{provincias gobernadas por el PJ}\}$$

$$A \cap B = \{\text{provincias con más del 40\% de pobreza Y gobernadas por el PJ}\}$$

#### 5. Complemento (Aᶜ)

<figure style="text-align:center; margin:1em 0;">
  <img src="media/complemento.png"  width="520">
  <figcaption><em>Union</em></figcaption>
</figure>

El complemento de A contiene **todos los elementos del universo U que NO pertenecen a A**.

$$A^c = \{x \in U \mid x \notin A\}$$

$$P(A^c) = 1 - P(A)$$

**Ejemplo:**
$$U = \{\text{todos los senadores nacionales}\} \quad |U| = 72$$
$$A = \{\text{senadores que apoyaron la ley de financiamiento educativo}\} \quad |A| = 45$$

$$A^c = \{\text{senadores que NO apoyaron la ley}\} \quad |A^c| = 27$$

#### 6. Diferencia (A − B)

La diferencia A − B contiene los elementos que **pertenecen a A pero no a B**.

$$A - B = \{x \mid x \in A \text{ y } x \notin B\}$$

**Ejemplo:**
$$A = \{\text{países de América Latina con sistema presidencialista}\}$$
$$B = \{\text{países de América Latina con reelección presidencial indefinida}\}$$

$$A - B = \{\text{países presidencialistas sin reelección indefinida}\}$$


#### 7. Diferencia simétrica (A △ B)

Contiene los elementos que pertenecen a A o a B
**pero no a ambos simultáneamente**.

$$A \triangle B = (A - B) \cup (B - A) = (A \cup B) - (A \cap B)$$

**Ejemplo:**
$$A = \{\text{legisladores que votaron la reforma laboral}\}$$
$$B = \{\text{legisladores que votaron la reforma tributaria}\}$$

$$A \triangle B = \{\text{legisladores que votaron exactamente una de las dos reformas}\}$$

```python
# Teoría de conjuntos con Python: el tipo `set` reproduce las operaciones
# unión (|), intersección (&), diferencia (-) y diferencia simétrica (^).

U = {"UCR", "PJ", "PRO", "FIT", "MID", "Libres del Sur"}   # universo
A = {"UCR", "PJ", "PRO"}                                    # partidos con +20 bancas

print("¿A es subconjunto de U? (A ⊆ U):", A.issubset(U))

# Cardinalidad de la unión:  |A ∪ B| = |A| + |B| − |A ∩ B|
nA, nB, nAB = 180, 140, 90
print("Unión por cardinalidad |A∪B| =", nA + nB - nAB)     # 230

# Operaciones elemento a elemento sobre conjuntos de legisladores
reforma_laboral    = {"Díaz", "Gómez", "Ruiz", "Sosa"}
reforma_tributaria = {"Gómez", "Ruiz", "Pérez"}

print("Unión (votaron al menos una):   ", reforma_laboral | reforma_tributaria)
print("Intersección (votaron ambas):   ", reforma_laboral & reforma_tributaria)
print("Diferencia (solo laboral):      ", reforma_laboral - reforma_tributaria)
print("Dif. simétrica (exactamente una):", reforma_laboral ^ reforma_tributaria)

# Complemento de A respecto del universo U:  Aᶜ = U − A
print("Complemento Aᶜ = U − A:         ", U - A)
```

### Definición axiomática de probabilidad: algunas reglas

- No negatividad: $P (A) \geq 0, \forall A \subseteq \Omega$

- Las probabilidades suman 1: $P (\Omega) = 1$

- Las probabilidades están entre 0 y 1: $0 \leq P(A) \leq 1$

- Probabilidad de un subconjunto: Siendo $A$ y $B$ eventos en $\Omega$ ($A, B \subseteq \Omega$), si $B \subseteq A \rightarrow P(B) \leq P(A)$ 

- Regla de la suma para eventos mutuamente excluyentes: Si P(A) = 0.20 y P(B) = 0.35
P(A o B) = 0.55


### Regla de la Suma para Eventos Mutuamente Excluyentes

Cuando dos eventos pueden ocurrir al mismo tiempo, sus probabilidades simplemente se suman: 

- Si $P(A) = 0.20$ y $P (B) = 0.35$: $$ P(A \text { o } B) = 0.55$$

### Regla de la Suma para Eventos NO Mutuamente Excluyentes

Si los sucesos no pueden ocurrir al mismo tiempo, la regla de la adición se formula *restando la probabilidad de que ocurran juntos* para no duplicar esa área: $$ P (A \text { o } B) = P(A) + P(B) - P(A \text { y } B)$$

Expresado formalmente mediante la teoría de conjuntos: $$ P (A \cup B) = P(A) + P(B) - P(A \cap B)$$




Donde: 
$P(A \cap B)$ o           $P(A \text { y } B)$
denota la probabilidad de que $A$ y $B$ ocurran al mismo tiempo (**INTERSECCIÓN**)

### Eventos equiprobables

#### Definición

Un espacio muestral es **equiprobable** cuando todos sus resultados posibles tienen exactamente **la misma probabilidad de ocurrir**.

Si el espacio muestral tiene $n$ resultados posibles y todos son equiprobables, la probabilidad de cada uno es:

$$P(\omega_i) = \frac{1}{n} \quad \text{para todo } i = 1, 2, \ldots, n$$

Y la probabilidad de cualquier evento A con $k$ resultados favorables es:

$$P(A) = \frac{\text{casos favorables}}{\text{casos posibles}} = \frac{k}{n}$$

Esta es la **definición clásica de probabilidad**.



#### Ejemplo — Muestreo aleatorio de ciudadanos

Se selecciona **1 ciudadano** al azar de una lista de 8 integrantes de un grupo focal sobre confianza institucional.

$$\Omega = \{\text{Ana}, \text{Bruno}, \text{Carlos}, \text{Diana},
\text{Elena}, \text{Franco}, \text{Gisela}, \text{Hernán}\}$$
$$n = 8 \quad \Rightarrow \quad P(\omega_i) = \frac{1}{8} = 0{,}125$$

| Ciudadano | Edad | Aprueba gestión | Voto 2023 |
|:----------|:----:|:---------------:|:---------:|
| Ana       | 34   | Sí              | Oficialismo |
| Bruno     | 52   | No              | Oposición |
| Carlos    | 27   | No              | Oposición |
| Diana     | 45   | Sí              | Oficialismo |
| Elena     | 61   | Sí              | Oficialismo |
| Franco    | 38   | No              | Oposición |
| Gisela    | 29   | Sí              | Oposición |
| Hernán    | 55   | No              | Oficialismo |

**Evento D** = "El ciudadano seleccionado aprueba la gestión"

$$D = \{\text{Ana}, \text{Diana}, \text{Elena}, \text{Gisela}\} \quad k = 4$$

$$P(D) = \frac{4}{8} = 0{,}50$$

**Evento E** = "El ciudadano votó al oficialismo en 2023"

$$E = \{\text{Ana}, \text{Diana}, \text{Elena}, \text{Hernán}\} \quad k = 4$$

$$P(E) = \frac{4}{8} = 0{,}50$$

**Intersección D ∩ E** = "Aprueba la gestión Y votó al oficialismo"

$$D \cap E = \{\text{Ana}, \text{Diana}, \text{Elena}\} \quad k = 3$$

$$P(D \cap E) = \frac{3}{8} = 0{,}375$$

**Unión D ∪ E** = "Aprueba O votó al oficialismo (o ambas)"

$$P(D \cup E) = P(D) + P(E) - P(D \cap E) = 0{,}50 + 0{,}50 - 0{,}375 = 0{,}625$$

#### Propiedad clave: verificación de equiprobabilidad

Para que un espacio sea equiprobable se deben cumplir **dos condiciones simultáneamente**:

$$\sum_{i=1}^{n} P(\omega_i) = 1 \qquad \text{y} \qquad P(\omega_i) = \frac{1}{n} \; \forall i$$

En el Ejemplo 3:

$$8 \times \frac{1}{8} = 1 \checkmark$$

Si las probabilidades fueran distintas entre ciudadanos (por ejemplo, porque algunos tienen mayor probabilidad de ser contactados por vivir más cerca), el espacio **dejaría de ser equiprobable** y la fórmula clásica de Laplace ya no sería válida.

```python
from fractions import Fraction

# Espacio equiprobable: P(ω) = 1/n ; P(A) = k/n  (definición clásica de Laplace)

# Ejemplo 1 — sortear 1 diputado entre 5 ; A = {Fernández, García}
P_A = Fraction(2, 5)
print("Ej.1  P(A) =", P_A, "=", float(P_A), " | P(Aᶜ) =", 1 - P_A)

# Ejemplo 2 — sortear 1 provincia entre 24 ; B patagónicas, C grandes (B ∩ C = ∅)
P_B, P_C = Fraction(5, 24), Fraction(4, 24)
print("Ej.2  P(B) =", P_B, "≈", round(float(P_B), 3),
      "| P(B∪C) =", P_B + P_C, "=", float(P_B + P_C))

# Ejemplo 3 — grupo focal de 8 ; D = aprueba, E = oficialismo (D ∩ E no vacío)
P_D, P_E, P_DE = Fraction(4, 8), Fraction(4, 8), Fraction(3, 8)
print("Ej.3  P(D∪E) = P(D)+P(E)−P(D∩E) =", P_D + P_E - P_DE,
      "=", float(P_D + P_E - P_DE))
```

```python
import numpy as np

# ---------------------------------------------------------------
# Problema del cumpleaños
# ¿Qué tan probable es que en un grupo AL MENOS 2 personas
# compartan día de cumpleaños?  (365 días equiprobables)
#
# Truco (como en la diapositiva): INVERTIR el problema y trabajar
# con el COMPLEMENTO. Es más fácil calcular P(todos cumplen en
# días DISTINTOS) y después restar de 1:
#     P(al menos 2 comparten) = 1 - P(todos distintos)
# ---------------------------------------------------------------

def prob_todos_distintos(n, dias=365):
    # P(distintos) = 365/365 · 364/365 · 363/365 · … · (365-(n-1))/365
    # persona 1: cualquier día; persona 2 evita 1 día ya usado;
    # persona k evita (k-1) días -> quedan (365-(k-1)) días libres
    p = 1.0
    for k in range(n):
        p *= (dias - k) / dias
    return p

# Paso a paso, igual que en la diapositiva:
print(" n   P(todos distintos)   P(al menos 2 comparten)")
for n in [2, 3, 5, 10, 23, 30]:
    p_dist = prob_todos_distintos(n)
    print(f"{n:>2}        {p_dist:6.4f}               {1 - p_dist:6.4f}")

# El caso de la diapositiva: grupo de 30 personas
n = 30
p_dist_30 = prob_todos_distintos(n)
print(f"\nGrupo de {n} personas:")
print(f"  P(todos en días distintos) = {p_dist_30:.4f}   (≈ 0.30)")
print(f"  P(al menos 2 comparten)    = {1 - p_dist_30:.4f}   (≈ 0.70)  ->  ¡70%!")

# ---------------------------------------------------------------
# Verificación empírica por simulación (Monte Carlo)
# Generamos muchos grupos de 30 cumpleaños al azar y contamos en
# qué proporción hay al menos un día repetido.
# ---------------------------------------------------------------
rng = np.random.default_rng(88)          # semilla fija -> reproducible
n_sim = 100_000
grupos = rng.integers(1, 366, size=(n_sim, 30))   # 100.000 grupos de 30 días
grupos.sort(axis=1)                                # ordenar cada fila
comparten = (np.diff(grupos, axis=1) == 0).any(axis=1)  # ¿días iguales contiguos?
print(f"\nSimulación de {n_sim:,} grupos de 30:")
print(f"  proporción con cumpleaños compartido ≈ {comparten.mean():.4f}  (coincide con el 0.70 teórico)")
```

### Probabilidad Condicional  
La probabilidad condicional mide la probabilidad de que ocurra un evento A **dado que ya ocurrió** un evento B. Actualiza la probabilidad de A restringiendo el espacio muestral al evento B que se tomó como condición.

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)} \qquad \text{con } P(B) > 0$$

  <figcaption><em></em></figcaption>
</figure>

Supongamos que dado $\Omega = \{1,2,3,4,5,6\}$, conocemos que ocurrió el evento $B$:

$$B = \{1,3,5\} \qquad\qquad A = \{1,2,3\}$$

¿Cuál es la *probabilidad de ocurrencia* del evento $A$?

Inicialmente:

<figure style="text-align:center; margin:1em 0;">
  <img src="media/diceven.png"  width="320">
  <figcaption><em></em></figcaption>
</figure>

$$P(A) = \frac{3}{6} = \frac{1}{2} \qquad\qquad P(B) = \frac{3}{6} = \frac{1}{2}$$

Y sabiendo que ocurrió $B$:

$$
\begin{aligned}
P(A \mid B) &= \;? \\[4pt]
P(B) &= \tfrac{1}{2} \\[4pt]
P(A \mid B) &= P(A \cap B)\,\mid B \\[4pt]
            &= \frac{P(A \cap B)}{P(B)} = \frac{1/6}{3/6} = \frac{1}{3}
\end{aligned}
$$

#### Ley de la Multiplicación

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)} \qquad\Longrightarrow\qquad P(A \cap B) = P(A \mid B) \cdot P(B)$$

### Propiedades

- $0 \leq P(A \mid B) \leq 1$
- $P(A^c \mid B) = 1 - P(A \mid B)$
- Si A y B son **independientes**: $P(A \mid B) = P(A)$
  (conocer B no modifica la probabilidad de A)


#### Ley de la Probabilidad Total

Permite determinar la probabilidad **incondicional** de un evento a partir de probabilidades condicionales.

Dado un evento $A$ y una partición del espacio muestral en $n$ partes $B_1, B_2, \ldots, B_n$, la *probabilidad total de $A$* es:

$$P(A) = P(A \mid B_1)\,P(B_1) + P(A \mid B_2)\,P(B_2) + \cdots + P(A \mid B_n)\,P(B_n)$$

o, en forma compacta:

$$P(A) = \sum_{i=1}^{n} P(A \mid B_i)\,P(B_i)$$

Los $B_i$ $(i = 1, \ldots, n)$ conforman la totalidad del espacio muestral y son **mutuamente excluyentes**:

$$\bigcup_{i=1}^{n} B_i = \Omega \qquad\qquad B_i \cap B_j = \emptyset \quad (i \neq j)$$


### Probabilidad condicional y ley de la probabilidad total. Ejemplos


#### Ejemplos de probabilidad condicional en Ciencia Política

#### Ejemplo 1 — Voto e ideología

Se encuestan 500 ciudadanos. Se registran dos variables: ideología (Progresista / Conservadora) y voto (Oficialismo / Oposición).

|                  | Oficialismo | Oposición | Total |
|:-----------------|:-----------:|:---------:|:-----:|
| **Progresista**  | 180         | 70        | 250   |
| **Conservadora** | 60          | 190       | 250   |
| **Total**        | 240         | 260       | 500   |

**¿Cuál es la probabilidad de votar al oficialismo dado que el ciudadano es progresista?**

$$P(\text{Ofic} \mid \text{Prog}) = \frac{P(\text{Ofic} \cap \text{Prog})}{P(\text{Prog})}
= \frac{180/500}{250/500} = \frac{180}{250} = 0{,}72$$

**¿Y dado que es conservador?**

$$P(\text{Ofic} \mid \text{Cons}) = \frac{60}{250} = 0{,}24$$

> La ideología **condiciona fuertemente** el voto:
> entre progresistas hay 72% de apoyo al oficialismo,
> entre conservadores solo 24%. Las variables **no son independientes**.

**Verificación de independencia:**

Si fueran independientes debería cumplirse:
$P(\text{Ofic} \mid \text{Prog}) = P(\text{Ofic})$

$$P(\text{Ofic}) = \frac{240}{500} = 0{,}48 \neq 0{,}72$$

No se cumple → **dependencia estadística confirmada**.

#### 3. Ley de la probabilidad total

La ley de la probabilidad total permite calcular la probabilidad de un evento A cuando el espacio muestral se puede **particionar** en eventos mutuamente excluyentes y exhaustivos $B_1, B_2, \ldots, B_k$.

#### Condiciones de la partición

$$B_i \cap B_j = \emptyset \quad \forall i \neq j
\qquad \text{y} \qquad \bigcup_{i=1}^{k} B_i = \Omega$$

#### Fórmula

$$P(A) = \sum_{i=1}^{k} P(A \mid B_i) \cdot P(B_i)$$

Cada término $P(A \mid B_i) \cdot P(B_i)$ es la contribución del grupo $B_i$ 
a la probabilidad total de A, ponderada por el tamaño relativo del grupo.

#### 4. Ejemplos de la ley de la probabilidad total

#### Ejemplo 4 — Probabilidad total de votar al oficialismo

La población se divide en tres grupos ideológicos (partición de Ω) con sus respectivas probabilidades:

| Grupo (Bᵢ)       | P(Bᵢ) | P(Ofic \| Bᵢ) |
|:-----------------|:-----:|:-------------:|
| Progresista (B₁) | 0,40  | 0,72          |
| Centrista (B₂)   | 0,35  | 0,45          |
| Conservador (B₃) | 0,25  | 0,24          |

$$P(\text{Ofic}) = P(\text{Ofic} \mid B_1)\cdot P(B_1)
+ P(\text{Ofic} \mid B_2)\cdot P(B_2)
+ P(\text{Ofic} \mid B_3)\cdot P(B_3)$$

$$P(\text{Ofic}) = 0{,}72 \times 0{,}40
+ 0{,}45 \times 0{,}35
+ 0{,}24 \times 0{,}25$$

$$P(\text{Ofic}) = 0{,}288 + 0{,}1575 + 0{,}060 = \mathbf{0{,}5055}$$

> El oficialismo obtendría aproximadamente el **50,6% de los votos**
> si la composición ideológica del electorado fuera esa.
> La ley de la probabilidad total "pondera" el apoyo de cada grupo
> por su peso en el electorado.

```python
import pandas as pd

# --- Probabilidad condicional a partir de una tabla de contingencia (Ej.1) ---
# Ideología (Progresista / Conservadora) x Voto (Oficialismo / Oposición), N = 500
tabla = pd.DataFrame({"Oficialismo": [180, 60], "Oposicion": [70, 190]},
                     index=["Progresista", "Conservadora"])
N = tabla.values.sum()

# P(Ofic | Prog) = n(Ofic ∩ Prog) / n(Prog)
P_ofic_prog = tabla.loc["Progresista", "Oficialismo"] / tabla.loc["Progresista"].sum()
P_ofic_cons = tabla.loc["Conservadora", "Oficialismo"] / tabla.loc["Conservadora"].sum()
P_ofic      = tabla["Oficialismo"].sum() / N
print(f"P(Ofic|Prog) = {P_ofic_prog:.2f}")
print(f"P(Ofic|Cons) = {P_ofic_cons:.2f}")
print(f"P(Ofic)      = {P_ofic:.2f}  ->  P(Ofic|Prog) ≠ P(Ofic): variables dependientes")

# --- Ley de la probabilidad total (Ej.4): P(A) = Σ P(A|Bi)·P(Bi) ---
grupos = pd.DataFrame({"P_B": [0.40, 0.35, 0.25],
                       "P_ofic_dado_B": [0.72, 0.45, 0.24]},
                      index=["Progresista", "Centrista", "Conservador"])
grupos["contribucion"] = grupos["P_B"] * grupos["P_ofic_dado_B"]
print("\nLey de la probabilidad total (Ej.4):")
print(grupos)
print(f"P(Ofic) = suma de contribuciones = {grupos['contribucion'].sum():.4f}")
```

### Aplicación de la Ley de Probabilidad Total a Encuestas Sensibles

Para realizar una encuesta se le pide a los participantes que saquen al azar (y observen **en privado**) una carta de un mazo, que determina qué pregunta deben responder:

- **40%** → pregunta **inocua** (PI): *¿Usted ha nacido en año par?*
- **60%** → pregunta **de interés** (Pint): *¿Alguna vez ha consumido drogas ilegales?*

Cada encuestado responde **sí / no**, pero no revela qué pregunta contestó. Entonces, ¿podemos conocer el porcentaje de la muestra que ha consumido drogas ilegales?

Por la **ley de la probabilidad total**, la proporción de respuestas "sí" se descompone según la pregunta que le tocó a cada participante:

$$P(\text{sí}) = P(\text{sí} \mid \text{PI})\cdot P(\text{PI}) + P(\text{sí} \mid \text{Pint})\cdot P(\text{Pint})$$

$$P(\text{sí}) = P(\text{sí} \mid \text{PI})\cdot 0{,}40 + P(\text{sí} \mid \text{Pint})\cdot 0{,}60$$

Podemos aproximar la proporción de personas nacidas en año par, de modo que suponemos $P(\text{sí} \mid \text{PI}) = 0{,}5$. Reemplazando:

$$P(\text{sí}) = 0{,}5 \cdot 0{,}40 + P(\text{sí} \mid \text{Pint})\cdot 0{,}60$$

Y despejamos la incógnita de interés $P(\text{sí} \mid \text{Pint})$:

$$P(\text{sí} \mid \text{Pint}) = \frac{P(\text{sí}) - 0{,}5 \cdot 0{,}40}{0{,}60}$$

donde $P(\text{sí})$ es la proporción **observada** en la muestra:

$$P(\text{sí}) = \frac{\text{cantidad de respuestas afirmativas}}{\text{total de participantes}}$$



#### Eventos independientes

Dos eventos son independientes entre sí cuando conocer información sobre la probabilidad de uno no es relevante para estimar la probabilidad del otro. Esto supone que: 

$$ P(A|B) = P(A)$$

¿Qué ejemplos de eventos independientes se pueden pensar?

Ejemplo: 

- A= mañana va a aumentar el subte

- B= Esta clase terminará a las 22

**podemos suponer que ambos eventos son independientes, en tanto que el aumento del subte no aumenta ni disminuye las probabilidades de que la clase termine a las 22 y viceversa**

Con eventos independientes, un pasaje de eventos que se puede hacer es el siguiente:

$$ 
P(A) = P(A \cap B) / P(B)

P(A \cap B) = P(A) * P(B)
$$

Si se cumple esta última igualdad entonces se trata de eventos independientes, de lo contrario, se trata de eventos dependientes.

### Teorema de Bayes

**Probabilidad subjetiva:**

Ya no hablamos de probabilidad frecuentista o equiprobabilidad, sino que la probabilidad de un evento se relaciona con las creencias del observador. La probabilidad se actualiza con la obtención de nueva información.

**Aplicaciones:**

- Medicina
- Juicios, criminología
- Economía y finanzas
- Marketing, diseño web

Frecuentemente sucede que se quiere calcular $P(A \mid B)$, cuando en realidad lo que se conoce es $P(B \mid A)$, $P(A)$ o $P(A^c)$.

### Teorema de Bayes: motivación

Ejemplo de uso de tests médicos para detectar una enfermedad rara:

Dada la existencia de tests médicos poco específicos (alto % de falsos positivos), no se suele someter a toda la población a testeos frecuentes de todas las enfermedades. Más bien se utilizan para confirmación de pacientes con síntomas.

**FALSOS POSITIVOS: ¿Cuál es la probabilidad de que una persona esté sana siendo que obtuvo un test positivo de una enfermedad X?**

$$P(A \mid B)$$

Donde: $P(A) = \text{Sano}$, $P(B) = \text{+}$

### Teorema de Bayes: motivación. Ejemplo numérico

**Supuestos:**

- Una enfermedad que solo afecta al $1\%$ de la población.
- Si una persona está enferma, el test le dará positivo con probabilidad $0{,}98$.
- Si la persona está sana, el test dará positivo con una probabilidad de $0{,}05$ ($0{,}99$ de probabilidad de estar sano).

Retomando la pregunta anterior con estos datos concretos:

**¿Cuál es la probabilidad de que una persona esté sana siendo que obtuvo un test positivo de una enfermedad X?**

$$P(A \mid B)$$

Donde: $P(A) = \text{la persona está sana}$, $P(B) = \text{la persona dio +}$

¿Y si en cambio conocemos $P(B \mid A)$?

### Construcción del teorema de Bayes

El teorema de Bayes es una *igualdad matemática* resultante de combinar:

- Ley de la multiplicación
- Probabilidad conjunta
- Ley de la probabilidad total

### Probabilidad condicional: Ley de la multiplicación

$$P(A \mid B) = \dfrac{P(A \cap B)}{P(B)}$$

$$P(A \cap B) = P(A \mid B) \cdot P(B)$$

**Simetría de la probabilidad conjunta:**

$$P(A \cap B) = P(B \cap A)$$

### Probabilidad condicional: Ley de la probabilidad total

Permite determinar la probabilidad incondicional de un evento dependiendo de probabilidades condicionales.

Dado un evento $A$ y una partición del espacio muestral en $B_1, B_2, \ldots, B_n$, la *probabilidad total* de $A$ es:

$$P(A) = P(A \mid B_1) \cdot P(B_1) + P(A \mid B_2) \cdot P(B_2) + \cdots + P(A \mid B_n) \cdot P(B_n)$$

Notemos que los $B_i$, $i = 1, \ldots, n$, conforman la totalidad del espacio muestral y son **mutuamente excluyentes** (no presentan elementos en común).

Resolvamos el problema

| Dato | Valor | Nombre técnico |
|:--|:--:|:--|
| Prevalencia de la enfermedad en la población | 1% | Probabilidad a priori |
| P(Test + \| Enfermo) | 0,98 | Sensibilidad del test |
| P(Test + \| Sano) | 0,05 | Tasa de falsos positivos (1 − especificidad) |

### Definición de eventos

$$A = \text{"La persona está enferma"}$$
$$A^c = \text{"La persona está sana"}$$
$$B = \text{"El test da positivo"}$$

### Datos del problema

$$P(A) = 0{,}01 \qquad P(A^c) = 0{,}99$$
$$P(B \mid A) = 0{,}98 \qquad P(B \mid A^c) = 0{,}05$$


### Paso 1: Probabilidad total de dar positivo P(B)

$$P(B) = P(B \mid A) \cdot P(A) + P(B \mid A^c) \cdot P(A^c)$$

$$P(B) = 0{,}98 \times 0{,}01 + 0{,}05 \times 0{,}99$$

$$P(B) = 0{,}0098 + 0{,}0495 = 0{,}0593$$

De cada 100 personas testeadas, aproximadamente **5,93 darán positivo**, aunque la mayoría de ellas estarán sanas.

### Paso 2: Aplicar el teorema de Bayes

**¿Cuál es la probabilidad de estar enfermo dado que el test dio positivo?**

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}
= \frac{0{,}98 \times 0{,}01}{0{,}0593}
= \frac{0{,}0098}{0{,}0593} \approx 0{,}1653$$

$$\boxed{P(\text{Enfermo} \mid \text{Test}+) \approx 16{,}53\%}$$

### Paso 3: Interpretación contraintuitiva

A pesar de que el test tiene una sensibilidad del 98%, **solo el 16,5% de los que dan positivo están realmente enfermos**. El 83,5% restante son **falsos positivos**.

¿Por qué? Porque la enfermedad es muy poco frecuente (1%).
La gran mayoría de los testeados son sanos, y aunque
la tasa de falsos positivos es baja (5%), al aplicarse
sobre una base enorme de sanos produce muchos positivos espurios.

| Resultado | Cantidad (sobre 10.000 personas) |
|:--|:--:|
| Enfermos con test + (verdaderos positivos) | $10.000 \times 0{,}01 \times 0{,}98 = \mathbf{98}$ |
| Sanos con test + (falsos positivos) | $10.000 \times 0{,}99 \times 0{,}05 = \mathbf{495}$ |
| **Total con test +** | **593** |
| **Enfermos entre los positivos** | **98 / 593 ≈ 16,5%** |

### Paso 4: Efecto de la prevalencia sobre el valor predictivo

La probabilidad a posteriori cambia drásticamente según cuán común sea la enfermedad en la población:

| Prevalencia P(A) | P(Enfermo \| Test +) |
|:--:|:--:|
| 0,001 (0,1%) | 1,9% |
| 0,01  (1%)   | 16,5% |
| 0,05  (5%)   | 50,8% |
| 0,10  (10%)  | 68,6% |
| 0,30  (30%)  | 89,3% |
| 0,50  (50%)  | 95,1% |

> **Conclusión:** un test con buena sensibilidad puede tener
> un valor predictivo positivo muy bajo si la enfermedad
> es rara. El teorema de Bayes formaliza por qué
> el screening masivo en poblaciones de bajo riesgo
> genera una avalancha de falsos positivos.

```python
# Teorema de Bayes aplicado a un test diagnóstico (enfermedad rara).
P_A         = 0.01    # prevalencia (prior): P(Enfermo)
P_B_dado_A  = 0.98    # sensibilidad:        P(Test+ | Enfermo)
P_B_dado_Ac = 0.05    # falsos positivos:    P(Test+ | Sano)
P_Ac        = 1 - P_A

# Paso 1 — Ley de la probabilidad total: P(Test+)
P_B = P_B_dado_A * P_A + P_B_dado_Ac * P_Ac
print(f"P(Test+)            = {P_B:.4f}")

# Paso 2 — Teorema de Bayes: P(Enfermo | Test+)
P_A_dado_B = (P_B_dado_A * P_A) / P_B
print(f"P(Enfermo | Test+)  = {P_A_dado_B:.4f}  ({P_A_dado_B*100:.1f}%)")

# Paso 3 — Verificación con frecuencias naturales sobre 10.000 personas
N = 10_000
VP = round(N * P_A  * P_B_dado_A)     # verdaderos positivos
FP = round(N * P_Ac * P_B_dado_Ac)    # falsos positivos
print(f"\nSobre {N:,} personas: {VP} verdaderos positivos y {FP} falsos positivos")
print(f"Enfermos entre los positivos = {VP}/{VP+FP} = {VP/(VP+FP):.4f}")

# Paso 4 — Efecto de la prevalencia sobre el valor predictivo positivo (VPP)
print("\nPrevalencia P(A)  ->  P(Enfermo | Test+):")
for prev in [0.001, 0.01, 0.05, 0.10, 0.30, 0.50]:
    vpp = (P_B_dado_A * prev) / (P_B_dado_A * prev + P_B_dado_Ac * (1 - prev))
    print(f"  {prev:<6} -> {vpp*100:5.1f}%")
```
