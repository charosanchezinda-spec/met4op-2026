## Conceptos básicos de la estadística inferencial

### Diferencia entre estadística descriptiva e inferencial

- Población: Conjunto total de elementos que se estudian
- Muestra: Subconjunto representativo de la población
- Muestreo Aleatorio: Método de selección de la muestra que asegura que cada individuo tiene la misma probabilidad de ser elegido, reduciendo sesgos.

### Parámetros y estimadores
- Parámetro poblacional: Valor numérico que describe una característica de la población
- Estadístico muestral: Valor numérico calculado a partir de la muestra, usado para estimar el parámetro poblacional

### Ejemplos

##### 1. Población

La población es el conjunto **completo** de unidades estadísticas
sobre las que se quiere hacer una afirmación.
Es el universo de referencia del estudio.

> La definición de la población depende del objeto de investigación:
> no es lo mismo estudiar "los argentinos" que
> "los legisladores nacionales" o "los partidos provinciales".

| Pregunta de investigación |
|:-------------------------- |
| ¿Cuántos argentinos aprueban la gestión presidencial? |
| ¿Qué porcentaje de legisladores nacionales apoyó la reforma previsional? |
| ¿Cuántos municipios bonaerenses tienen población de más de 100.000 habitantes? |
| ¿Qué proporción de partidos provinciales sobrevivió más de dos elecciones? |

> **Nota clave:** cuando la población es pequeña y accesible
> (como los 257 diputados), es posible relevarla completamente
> y obtener el parámetro directamente, sin necesidad de muestreo.
> Cuando es grande (el padrón electoral), el muestreo es indispensable.

##### 2. Muestra

La muestra es un **subconjunto** de la población,
seleccionado de forma tal que sea representativo
de las características del universo completo.

$$n < N$$

El tamaño de la muestra (n) es siempre menor que
el tamaño de la población (N).

**Estudio:** aprobación presidencial en Argentina.

- **Población (N):** ~33 millones de ciudadanos habilitados
- **Muestra (n):** 1.200 ciudadanos seleccionados aleatoriamente

| ID | Provincia    | Edad | Sexo | Aprueba gestión | Confianza inst. (0–10) |
|:--:|:-------------|:----:|:----:|:---------------:|:----------------------:|
| 1  | Buenos Aires | 34   | F    | Sí              | 6                      |
| 2  | Córdoba      | 52   | M    | No              | 3                      |
| 3  | Santa Fe     | 27   | F    | No              | 2                      |
| 4  | Mendoza      | 45   | M    | Sí              | 7                      |
| 5  | Tucumán      | 61   | F    | No              | 4                      |
| …  | …            | …    | …    | …               | …                      |
| 1200 | Neuquén  | 38   | M    | Sí              | 5                      |

> La muestra no es cualquier subconjunto: debe ser
> seleccionada con un procedimiento que garantice
> representatividad. De lo contrario, las inferencias
> sobre la población serán sesgadas.

##### 3. Muestreo aleatorio

El muestreo aleatorio es el procedimiento por el cual
cada unidad de la población tiene una **probabilidad
conocida y no nula** de ser seleccionada.
Es la condición que permite hacer inferencia válida.

#### Tipos principales y ejemplos políticos

##### Muestreo aleatorio simple

Cada ciudadano tiene exactamente la misma probabilidad
de ser seleccionado.

$$P(\text{selección}) = \frac{n}{N} = \frac{1200}{33.000.000} \approx 0{,}0036\%$$

**Ejemplo:** Se extrae al azar un listado de 1.200 DNI
del padrón electoral nacional para encuestar sobre
aprobación presidencial.

##### Muestreo estratificado

La población se divide en subgrupos (estratos) según
una variable relevante, y se selecciona aleatoriamente
dentro de cada estrato. Garantiza representación proporcional.

**Ejemplo:** Se divide el padrón en 24 distritos electorales
y se seleccionan encuestados en proporción al tamaño
del electorado de cada provincia.

| Provincia    | Electorado | % del total | n asignado |
|:-------------|:----------:|:-----------:|:----------:|
| Buenos Aires | 12.000.000 | 36,4%       | 437        |
| Córdoba      | 3.800.000  | 11,5%       | 138        |
| Santa Fe     | 3.500.000  | 10,6%       | 127        |
| Resto        | 13.700.000 | 41,5%       | 498        |
| **Total**    |**33.000.000**|**100%**   | **1.200**  |

##### Muestreo por conglomerados

Se seleccionan aleatoriamente grupos (conglomerados)
y se relevan todas las unidades dentro de ellos.

**Ejemplo:** Se sortean aleatoriamente 60 municipios
del país y se encuesta a todos los concejales de
esos municipios para estudiar representación de género
en los cuerpos legislativos locales.

##### Muestreo sistemático

Se elige un punto de inicio al azar y luego se selecciona
cada k-ésima unidad del listado.

$$k = \frac{N}{n} = \frac{33.000.000}{1.200} = 27.500$$

**Ejemplo:** Del padrón ordenado por DNI, se elige
aleatoriamente un número entre 1 y 27.500 como punto
de inicio y luego se selecciona cada 27.500 registros.

##### 4. Parámetro poblacional

El parámetro es una medida numérica que describe
una característica de la **población completa**.
Es un valor **fijo pero generalmente desconocido**,
porque relevarlo implicaría censar a toda la población.

Se denota con letras griegas:

| Parámetro | Símbolo | Descripción |
|:----------|:-------:|:------------|
| Proporción poblacional | π | Fracción de la población con cierta característica |
| Media poblacional | μ | Promedio de la población |
| Varianza poblacional | σ² | Dispersión en la población |
| Desvío estándar poblacional | σ | Raíz de la varianza poblacional |

**Ejemplo 1 — proporción:**
El verdadero porcentaje de ciudadanos argentinos que aprueba
la gestión del presidente en este momento.
Nadie lo conoce con exactitud: es el parámetro π.

**Ejemplo 2 — media:**
El verdadero promedio de confianza institucional
(medido en escala 0–10) de todos los argentinos mayores de 18 años.
Es el parámetro μ.

**Ejemplo 3 — varianza:**
La verdadera dispersión del porcentaje de votos obtenido
por un partido en todos los municipios del país.
Es el parámetro σ².

> Los parámetros existen en la realidad pero son inobservables
> en poblaciones grandes. La estadística inferencial permite
> estimarlos a partir de los estadísticos muestrales.

## 5. Estadístico muestral

El estadístico es una medida numérica calculada
a partir de los datos de la **muestra**.
Es observable, calculable, y se usa para **estimar**
el parámetro poblacional correspondiente.

Se denota con letras latinas o con "sombrero":

| Estadístico | Símbolo | Estima al parámetro |
|:------------|:-------:|:-------------------:|
| Proporción muestral | p̂ | π |
| Media muestral | x̄ | μ |
| Varianza muestral | s² | σ² |
| Desvío estándar muestral | s | σ |

##### Encuesta de aprobación presidencial

De los 1.200 encuestados:

- 624 aprueban la gestión del presidente
- 576 no aprueban

**Proporción muestral (estadístico):**

$$\hat{p} = \frac{624}{1200} = 0{,}52 \quad \Rightarrow \quad 52\%$$

**Media muestral de confianza institucional (estadístico):**

$$\bar{x} = \frac{\sum_{i=1}^{1200} x_i}{1200} = 4{,}3 \text{ puntos sobre 10}$$

**Desvío estándar muestral (estadístico):**

$$s = 2{,}1 \text{ puntos}$$

Estos tres valores son estadísticos: los calculamos
a partir de la muestra y los usamos para estimar
los parámetros reales de los 33 millones de argentinos.

#### Resumen integrador

| Concepto | Símbolo | ¿Observable? | Ejemplo |
|:---------|:-------:|:------------:|:--------|
| Población (N) | N | No completamente | ~33 millones de argentinos |
| Muestra (n) | n | Sí | 1.200 encuestados |
| Muestreo aleatorio | — | Sí (el procedimiento) | Sorteo del padrón electoral |
| Parámetro poblacional | π, μ, σ | No | Verdadero % de aprobación presidencial |
| Estadístico muestral | p̂, x̄, s | Sí | 52% de aprobación en la muestra |

La inferencia estadística es el puente entre
el estadístico (lo que medimos) y el parámetro
(lo que queremos conocer). Nunca alcanzamos el parámetro
con certeza absoluta, pero podemos estimarlo con
un margen de error calculable y una confianza conocida.

## Distribución muestral

Es una distribución teórica de probabilidades de los valores posibles de un estadístico, que se obtendría si tomáramos todas las muestras posibles de un tamaño fijo de la población

### Distribución normal: 
capacidad para modelar comportamientos donde la mayoría de las observaciones se concentran alrededor de un valor central, mientras que los valores extremos son menos frecuentes.

## Distribución muestral: estandarización
Frecuentemente se estandariza las distribuciones normales para llevarlas a un normal estandar o distribución t de student. Este procedimiento consiste en tomar la variable que se considera que sigue una distribución normal y *restarle su media y dividirla por su desviación estándar*


## Estadísticos: prueba de hipótesis
Estos estadísticos se usan para saber o determinar si la evidencia en datos es suficiente para rechazar la hipótesis nula $H_0$ en favor de la alternativa $H_1$.

### Prueba Z
La prueba Z se utiliza generalmente cuando se conoce la Desviación Estándar (DE) de la población o cuando el tamaño de la muestra es lo suficientemente grande como para que la estimación de la desviación estándar se muestre confiable. Se basa en la distribución normal estándar y se utiliza para hacer inferencias sobre la media poblacional *cuando los datos son aproximadamente normales* o el tamaño de la muestra es grande. 

### Prueba t de student
La prueba t de Student se utiliza cuando no se conoce la Desviación Estándar (DE) de la población y se debe estimar a partir de una muestra. También se utiliza cuando *el tamaño de una muestra es pequeño* y la distribución de los datos no es necesariamente normal. la distribución t tiene las colas mas gruesas que la distribución normal, lo que **refleja mayor incertidumbre** en la estimación del parámetro.

### Comparación entre estadístico y valor crítico
La regla general es comparar el valor absoluto del estadístico de prueba con el valor absoluto correspondiente al nivel de significación $\alpha$ previamente establecido.
- Si el estadístico de prueba calculado supera el valor crítico del valor absoluto (es decir, si cae en la *región de rechazo*), se rechaza la hipótesis nula.
- Si el estadístico de prueba calculado no supera el valor crítico (es decir, si cae fuera de la *región de rechazo*), no se rechaza la hipótesis nula.

***Es importante destacar que no rechazar la hipótesis nula no significa que la $H_0$ sea verdadera, sino que significa que no tiene suficiente evidencia para rechazarla con el nivel de significación elegido***

## 1. ¿Por qué estandarizar?

Las variables políticas se miden en escalas distintas: porcentajes de votos, índices de confianza de 0 a 10, cantidad de escaños, ingresos en pesos. Estandarizar convierte cualquier valor en una unidad común, sea el **valor z o valor t**, que expresa *cuántos desvíos estándar se aleja un resultado de la media*. Eso permite comparar resultados de distintas variables y ubicarlos en la distribución normal para hacer inferencia.

## 2. Estandarización con distribución normal (z)

Se usa cuando:
- el desvío estándar **poblacional (σ) es conocido**, o
- la muestra es **grande (n ≥ 30)**

### Fórmula para una observación

$$z = \frac{x - \mu}{\sigma}$$

### Fórmula para una media muestral

$$z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$$

### Fórmula para una proporción muestral

$$z = \frac{\hat{p} - \pi_0}{\sqrt{\dfrac{\pi_0(1-\pi_0)}{n}}}$$

### 3. Estandarización con distribución t de Student

Se usa cuando:
- el desvío estándar **poblacional (σ) es desconocido**
- se lo reemplaza por el **desvío muestral (s)**
- la muestra es **pequeña (n < 30)**

La distribución t es más achatada que la normal estándar y varía según los **grados de libertad (gl = n − 1)**. A medida que n crece, la t se aproxima a la normal.

### Fórmula para una media muestral

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$$

### Fórmula para diferencia de medias (dos grupos)

$$t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}$$

Donde:
- $\bar{x}$ = media muestral
- $\mu_0$ = valor del parámetro bajo H₀
- $s$ = desvío estándar muestral
- $n$ = tamaño de la muestra
- $gl = n - 1$ grados de libertad

```python
import numpy as np
from scipy import stats

# ── Ejemplo 1: estandarización z (n grande, σ poblacional conocido) ──
# Confianza institucional (0–10).   H0: μ = 5   H1: μ ≠ 5   (bilateral)
x_bar, mu0, sigma, n = 4.3, 5, 2.1, 1200
z = (x_bar - mu0) / (sigma / np.sqrt(n))
z_critico = stats.norm.ppf(1 - 0.05 / 2)                 # α = 0,05 bilateral
p_valor = 2 * (1 - stats.norm.cdf(abs(z)))
print("== Estandarización z ==")
print(f"z = {z:.2f}   z* = ±{z_critico:.2f}   p-value = {p_valor:.3g}")
print("Decisión:", "se rechaza H0" if abs(z) > z_critico else "no se rechaza H0")

# ── Ejemplo 2: estandarización t (n pequeño, σ desconocido) ──
# Acuerdo concejales (0–10).   H0: μ ≤ 5   H1: μ > 5   (cola derecha)
x_bar, mu0, s, n = 6.2, 5, 1.8, 18
gl = n - 1
t = (x_bar - mu0) / (s / np.sqrt(n))
t_critico = stats.t.ppf(1 - 0.05, gl)                    # α = 0,05 cola derecha
p_valor = 1 - stats.t.cdf(t, gl)
print("\n== Estandarización t ==")
print(f"t = {t:.3f}   t* = {t_critico:.3f} (gl={gl})   p-value = {p_valor:.3g}")
print("Decisión:", "se rechaza H0" if t > t_critico else "no se rechaza H0")
```

## Ejemplos: Estandarización y distribución t de Student

### 1. Ejemplo 1: estandarización z (n grande)

**Contexto:** Se encuestan n = 1.200 ciudadanos argentinos sobre confianza institucional (escala 0–10). De estudios anteriores se conoce σ = 2,1 (poblacional). La muestra da x̄ = 4,3. ¿Es significativamente distinto de 5?

$$H_0: \mu = 5 \qquad H_1: \mu \neq 5 \qquad \alpha = 0{,}05$$

**Estandarización:**

$$z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} = \frac{4{,}3 - 5}{2{,}1 / \sqrt{1200}} = \frac{-0{,}7}{2{,}1 / 34{,}64} = \frac{-0{,}7}{0{,}0606} \approx -11{,}55$$

**Valor crítico** para prueba bilateral con α = 0,05:

$$z^* = \pm 1{,}96$$

**Decisión:**

$$|z| = 11{,}55 > 1{,}96 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** La confianza institucional promedio de los argentinos es significativamente distinta de 5 puntos. Con una media muestral de 4,3, el resultado se ubica 11,55 desvíos estándar por debajo del valor de referencia: una diferencia estadísticamente muy significativa.

### 2. Ejemplo 2: estandarización t (n pequeño)

**Contexto:** Se entrevistan n = 18 concejales de municipios bonaerenses sobre su índice de acuerdo con el oficialismo
provincial (escala 0–10). La muestra da x̄ = 6,2 y s = 1,8. ¿Es el acuerdo significativamente mayor que 5?

$$H_0: \mu \leq 5 \qquad H_1: \mu > 5 \qquad \alpha = 0{,}05$$

**Grados de libertad:**

$$gl = n - 1 = 18 - 1 = 17$$

**Estandarización:**

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} = \frac{6{,}2 - 5}{1{,}8 / \sqrt{18}} = \frac{1{,}2}{1{,}8 / 4{,}243} = \frac{1{,}2}{0{,}4243} \approx 2{,}828$$

**Valor crítico** para prueba de cola derecha,
α = 0,05, gl = 17:

$$t^*_{(0{,}05;\,17)} = 1{,}740$$

**Decisión:**

$$t = 2{,}828 > 1{,}740 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** El índice de acuerdo de los concejales con el oficialismo provincial es significativamente mayor
que 5, el punto medio de la escala. Hay evidencia de alineamiento político con el gobierno provincial.

### 3. Ejemplo 3 — diferencia de medias t (dos grupos)

**Contexto:** Se compara el índice de confianza institucional entre ciudadanos de municipios con presupuesto participativo (Grupo 1, n₁ = 12) y sin presupuesto participativo (Grupo 2, n₂ = 14). ¿Difieren significativamente?

| | Grupo 1 (con PP) | Grupo 2 (sin PP) |
|:--|:--:|:--:|
| n | 12 | 14 |
| x̄ | 6,1 | 4,8 |
| s | 1,5 | 1,9 |

$$H_0: \mu_1 = \mu_2 \qquad H_1: \mu_1 \neq \mu_2 \qquad \alpha = 0{,}05$$

**Estandarización:**

$$t = \frac{6{,}1 - 4{,}8}{\sqrt{\dfrac{1{,}5^2}{12} + \dfrac{1{,}9^2}{14}}} = \frac{1{,}3}{\sqrt{\dfrac{2{,}25}{12} + \dfrac{3{,}61}{14}}} = \frac{1{,}3}{\sqrt{0{,}1875 + 0{,}2579}} = \frac{1{,}3}{\sqrt{0{,}4454}} = \frac{1{,}3}{0{,}6674} \approx 1{,}948$$

**Grados de libertad aproximados (Welch):**

$$gl \approx \frac{(0{,}1875 + 0{,}2579)^2}{\dfrac{0{,}1875^2}{11} + \dfrac{0{,}2579^2}{13}} = \frac{0{,}1985}{0{,}003196 + 0{,}005117} \approx \frac{0{,}1985}{0{,}008313} \approx 23$$

**Valor crítico** bilateral, α = 0,05, gl = 23:

$$t^*_{(0{,}025;\,23)} = \pm 2{,}069$$

**Decisión:**

$$|t| = 1{,}948 < 2{,}069 \quad \Rightarrow \quad \text{No se rechaza } H_0$$

**Interpretación:** No hay evidencia estadística suficiente para afirmar que el presupuesto participativo produce diferencias significativas en la confianza institucional con este tamaño muestral. Se requeriría una muestra mayor.

### 4. Comparación: estadístico t vs valor crítico t*

La regla de decisión en toda prueba t es:

| Prueba | Rechazar H₀ si… |
|:--|:--|
| Bilateral | $\|t\| > t^*_{\alpha/2,\,gl}$ |
| Cola derecha | $t > t^*_{\alpha,\,gl}$ |
| Cola izquierda | $t < -t^*_{\alpha,\,gl}$ |

### Tabla de valores críticos t más usados en Ciencia Política

| gl (n−1) | t* cola derecha α=0,05 | t* bilateral α=0,05 | t* bilateral α=0,01 |
|:--:|:--:|:--:|:--:|
| 5  | 2,015 | 2,571 | 4,032 |
| 10 | 1,812 | 2,228 | 3,169 |
| 17 | 1,740 | 2,110 | 2,898 |
| 23 | 1,714 | 2,069 | 2,807 |
| 29 | 1,699 | 2,045 | 2,756 |
| ∞  | 1,645 | 1,960 | 2,576 |

> A medida que gl crece, t* converge al valor z.
>Con gl = ∞, la t de Student es idéntica a la normal estándar.

### Resumen de los tres ejemplos

| Ejemplo | n | Estadístico | Valor crítico | Decisión |
|:--|:--:|:--:|:--:|:--:|
| Confianza inst. (z) | 1.200 | z = −11,55 | ±1,96 | Rechaza H₀ |
| Acuerdo concejales (t, cola derecha) | 18 | t = 2,828 | 1,740 | Rechaza H₀ |
| Presupuesto participativo (t, bilateral) | 12 y 14 | t = 1,948 | ±2,069 | No rechaza H₀ |

| Condición | Distribución |

## Nivel de significancia, nivel de confianza e intervalos

### 1. Nivel de significancia (α)

El nivel de significancia es la probabilidad máxima que el investigador acepta de **cometer un error al rechazar H₀ cuando en realidad es verdadera**.
Se fija *antes* de recolectar los datos.

$$\alpha = P(\text{rechazar } H_0 \mid H_0 \text{ es verdadera})$$

Es el umbral de tolerancia al error. Cuanto más chico, más exigente es la prueba.

### Valores convencionales en Ciencia Política

| α | Nivel de exigencia | Interpretación |
|:--:|:--:|:--|
| 0,10 | Bajo | Se acepta hasta un 10% de probabilidad de error |
| 0,05 | Estándar | El más usado en ciencias sociales |
| 0,01 | Alto | Se exige evidencia muy fuerte para rechazar H₀ |

### Ejemplo

**Pregunta:** ¿La aprobación presidencial es distinta del 50%?

$$H_0: \pi = 0{,}50 \qquad H_1: \pi \neq 0{,}50$$

Con **α = 0,05** el investigador acepta que, si rechaza H₀,
existe un 5% de probabilidad de estar equivocado
(es decir, de rechazar algo que en realidad era verdadero).

> En Ciencia Política se usa casi siempre α = 0,05
> porque las variables sociales tienen alta variabilidad
> natural y α = 0,01 resultaría demasiado restrictivo.

## 2. Nivel de confianza (1 − α)

El nivel de confianza es el complemento del nivel de significancia. Expresa la probabilidad de que el intervalo construido **contenga al verdadero parámetro**.

$$\text{Nivel de confianza} = 1 - \alpha$$

| α | Nivel de confianza | z crítico |
|:--:|:--:|:--:|
| 0,10 | 90% | 1,645 |
| 0,05 | 95% | 1,960 |
| 0,01 | 99% | 2,576 |

### Ejemplo

Con α = 0,05 → nivel de confianza = **95%**.

Esto significa: si se construyeran 100 intervalos a partir de 100 muestras distintas del padrón electoral, aproximadamente 95 de ellos contendrían el verdadero porcentaje de aprobación presidencial de toda la población.

## 3. Intervalo de confianza

Un intervalo de confianza es un rango de valores construido a partir del estadístico muestral, dentro del cual se estima que se encuentra el parámetro poblacional con un nivel de confianza dado. Un intervalo de Confianza del 95% implica que, si repitiésemos el experimento muchas veces y calculáramos el intervalo de confianza para *cada muestra*, aproximadamente el 95% de esos intervalos contendrían el valor real del parámetro. En otras palabras, hay un 95% de probabilidad de que el valor verdadero del parámetro se encuentre *dentro del intervalo estimado*

$$IC = \text{estadístico} \pm \text{margen de error}$$

## 4. Margen de error

El margen de error (E) es la mitad del ancho del intervalo. Indica cuánto puede alejarse el estadístico del parámetro real con el nivel de confianza elegido. Los márgenes de error suelen reportarse en encuestas e investigaciones y son muy importantes a la hora de extraer conclusiones de los parámetros presentados.

$$E = z_{\alpha/2} \cdot \sigma_{\hat{p}} \quad \text{(para proporciones)}$$

$$E = z_{\alpha/2} \cdot \frac{s}{\sqrt{n}} \quad \text{(para medias)}$$

### Ejemplo: aprobación presidencial

De una muestra de n = 1.200 ciudadanos argentinos, 624 aprueban la gestión del presidente.

$$\hat{p} = \frac{624}{1200} = 0{,}52$$

Con nivel de confianza del 95% → $z_{\alpha/2} = 1{,}96$

**Error estándar:**

$$\sigma_{\hat{p}} = \sqrt{\frac{0{,}52 \times 0{,}48}{1200}} = \sqrt{\frac{0{,}2496}{1200}} = \sqrt{0{,}000208} \approx 0{,}0144$$

**Margen de error:**

$$E = 1{,}96 \times 0{,}0144 \approx 0{,}0282 \approx 2{,}82\%$$

**Intervalo de confianza:**

$$IC(\pi) = 0{,}52 \pm 0{,}0282$$

$$IC(\pi) = [0{,}4918 \; ; \; 0{,}5482]$$

$$IC(\pi) = [49{,}18\% \; ; \; 54{,}82\%]$$

**Interpretación:** Con un 95% de confianza, el verdadero porcentaje de aprobación presidencial entre todos los argentinos se encuentra entre el **49,18% y el 54,82%**.

### Efecto del nivel de confianza sobre el ancho del intervalo

Usando los mismos datos (n=1.200, p̂=0,52):

| Nivel de confianza | z | Margen de error | Intervalo |
|:--:|:--:|:--:|:--|
| 90% | 1,645 | ±2,37% | [49,63% ; 54,37%] |
| 95% | 1,960 | ±2,82% | [49,18% ; 54,82%] |
| 99% | 2,576 | ±3,71% | [48,29% ; 55,71%] |

A mayor nivel de confianza, mayor margen de error y más ancho el intervalo. Ganar certeza tiene un costo: menos precisión.

### Efecto del tamaño muestral sobre el margen de error

Con p̂ = 0,52 y nivel de confianza del 95%:

| n | Margen de error |
|:--:|:--:|
| 200  | ±6,92% |
| 400  | ±4,89% |
| 800  | ±3,46% |
| 1.200 | ±2,82% |
| 2.400 | ±2,00% |

Cuadruplicar el tamaño muestral reduce el margen de error a la mitad. El error disminuye con la raíz cuadrada de n, no linealmente.

```python
import numpy as np
from scipy import stats

# Intervalo de confianza para una proporción — aprobación presidencial.
# Muestra: n = 1.200 ; 624 aprueban  ->  p̂ = 0,52
p_hat, n = 0.52, 1200
se = np.sqrt(p_hat * (1 - p_hat) / n)                    # error estándar

print("IC para la proporción de aprobación (p̂ = 0,52 ; n = 1.200)\n")
for conf in (0.90, 0.95, 0.99):
    z = stats.norm.ppf(1 - (1 - conf) / 2)               # z crítico
    e = z * se                                           # margen de error
    lo, hi = p_hat - e, p_hat + e
    print(f"Confianza {int(conf*100)}%  ->  z = {z:.3f}   "
          f"margen = ±{e*100:.2f}%   IC = [{lo*100:.2f}% ; {hi*100:.2f}%]")
```
